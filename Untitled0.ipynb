```python
# --- CÉLULA 2: MINERAÇÃO SENSIBILIZADA ---

import pandas as pd
import numpy as np
from scipy.stats import ttest_ind
from IPython.display import display, Markdown

# 1. RECUPERANDO GRUPOS (Certifique-se que o bloco anterior rodou)
grupo_v = df_dados_brutos[df_dados_brutos['grupo_fenotipico'] == 'Hiper-Vigilante']
grupo_c = df_dados_brutos[df_dados_brutos['grupo_fenotipico'] == 'Controle']

print(f"🔬 Analisando {len(grupo_v)} amostras Hiper-Vigilantes contra {len(grupo_c)} Controles...")

# 2. MINERAÇÃO AMPLIADA (Vamos olhar 1000 genes para aumentar a chance)
genes_para_minerar = data_table.index[:1000]
resultados = []

for gene in genes_para_minerar:
    v_exp = grupo_v[gene].astype(float)
    c_exp = grupo_c[gene].astype(float)

    # Teste t
    t_stat, p_val = ttest_ind(v_exp, c_exp, equal_var=False, nan_policy='omit')

    # Fold Change (Razão de mudança)
    mean_v = v_exp.mean()
    mean_c = c_exp.mean()
    fc = (mean_v + 0.01) / (mean_c + 0.01)
    log2_fc = np.log2(fc)

    resultados.append({'Gene': gene, 'p_valor': p_val, 'Log2FC': log2_fc})

df_variantes = pd.DataFrame(resultados)

# 3. FILTRAGEM MAIS FLEXÍVEL (Ajustada para detectar assinaturas sutis)
# Reduzimos a exigência de Log2FC de 0.5 para 0.2 (mudança de ~15%)
limite_p = 0.05
limite_fc = 0.2

df_assinatura = df_variantes[
    (df_variantes['p_valor'] < limite_p) &
    (abs(df_variantes['Log2FC']) > limite_fc)
].copy()

# Se ainda assim der zero, vamos pegar os top 10 mais próximos da significância
if len(df_assinatura) == 0:
    print("⚠️ Filtros rigorosos não encontraram nada. Exibindo os 10 genes mais promissores...")
    df_assinatura = df_variantes.sort_values(by='p_valor').head(10)

display(Markdown(f"### 🎯 Resultado da Mineração: {len(df_assinatura)} Variantes Detectadas"))
display(df_assinatura.sort_values(by='p_valor').head(15))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# 1. Ampliando para 100 amostras
# O .iloc[:, :100] pega as primeiras 100 colunas (pessoas/testes)
amostras_100 = data_table.iloc[:, :100]
medias_100 = amostras_100.mean()

# 2. Definindo a "Régua" (Corte estatístico)
# Vamos considerar 'Hiper-Vigilante' quem está acima da média do grupo
media_geral = medias_100.mean()
limite_vigilancia = media_geral + (medias_100.std() * 0.5)

# 3. Definindo as cores: Vermelho (Alerta) e Azul (Calma)
cores = ['#FF3030' if m > limite_vigilancia else '#1E90FF' for m in medias_100]

# 4. Criando o Gráfico de Grande Escala
plt.figure(figsize=(20, 8))
plt.bar(medias_100.index, medias_100, color=cores, edgecolor='black', alpha=0.8)

# 5. Adicionando a linha de "Média Populacional"
plt.axhline(media_geral, color='black', linestyle='--', linewidth=2, label='Média de Prontidão')

# 6. Títulos e Legendas para o Congresso
plt.title("Rastreamento Genômico: Identificação de Fenótipos de Hiper-Vigilância (n=100)", fontsize=18)
plt.ylabel("Nível de Expressão Gênica Médio (Prontidão Molecular)", fontsize=14)
plt.xlabel("Amostras do Banco de Dados NCBI", fontsize=14)
plt.xticks(rotation=90, fontsize=7) # Rotaciona os IDs das 100 amostras

# Legenda personalizada
elementos_legenda = [
    Line2D([0], [0], color='#FF3030', lw=4, label='Grupo Hiper-Vigilante (Estado de Alerta)'),
    Line2D([0], [0], color='#1E90FF', lw=4, label='Grupo Controle (Estado Paradão)')
]
plt.legend(handles=elementos_legenda, fontsize=12)

plt.tight_layout()
plt.show()
```
