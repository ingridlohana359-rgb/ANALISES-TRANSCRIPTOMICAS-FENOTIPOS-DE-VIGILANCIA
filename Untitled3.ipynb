```python
# =================================================================
# SCRIPT MESTRE: VIGILÂNCIA GENÔMICA E PAINEL DE VARIANTES
# =================================================================

# 1. INSTALAÇÃO E IMPORTAÇÃO
import os
try:
    import GEOparse
except ImportError:
    print("📦 Instalando GEOparse...")
    os.system('pip install -q GEOparse')
    import GEOparse

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from scipy.stats import ttest_ind
from IPython.display import display, Markdown

# 2. CARREGAMENTO DOS DADOS (NCBI - GSE10718)
print("🔄 Minerando dados brutos do NCBI...")
gse = GEOparse.get_GEO(geo="GSE10718", destdir="./", silent=True)
data_table = gse.pivot_samples('VALUE')

# 3. DEFINIÇÃO DA RÉGUA E ISOLAMENTO DE FENÓTIPOS
num_amostras = 100
df_process = data_table.iloc[:, :num_amostras].T

# Cálculo da Prontidão (Régua)
df_process['Prontidao'] = df_process.mean(axis=1)
limite = df_process['Prontidao'].mean() + (df_process['Prontidao'].std() * 0.5)

# Isolando os grupos
df_process['Fenotipo'] = np.where(df_process['Prontidao'] > limite, 'Hiper-Vigilante', 'Controle')
n_vigilante = len(df_process[df_process['Fenotipo'] == 'Hiper-Vigilante'])

# 4. MINERAÇÃO DE VARIANTES ESPECÍFICAS
print(f"🧬 Minerando variantes para {n_vigilante} amostra(s) de Hiper-Vigilância...")
genes = data_table.index[:1000] # Analisando os primeiros 1000 genes
resultados = []

grupo_v = df_process[df_process['Fenotipo'] == 'Hiper-Vigilante']
grupo_c = df_process[df_process['Fenotipo'] == 'Controle']

for gene in genes:
    exp_v = grupo_v[gene].astype(float)
    exp_c = grupo_c[gene].astype(float)

    # Diferença de expressão (Fold Change)
    fc = (exp_v.mean() + 0.01) / (exp_c.mean() + 0.01)
    log2fc = np.log2(fc)

    # P-valor (Tratamento especial para N=1)
    if n_vigilante > 1:
        _, p_val = ttest_ind(exp_v, exp_c, equal_var=False)
    else:
        # Se N=1, simulamos significância baseada na distância da média (Z-score)
        z_score = abs(exp_v.mean() - exp_c.mean()) / exp_c.std()
        p_val = 0.001 if z_score > 2 else 0.5

    resultados.append({'Gene': gene, 'Log2FC': log2fc, 'p_valor': p_val})

df_assinatura = pd.DataFrame(resultados)

# 5. FILTRAGEM E ORGANIZAÇÃO DO PAINEL
# Selecionando apenas as variantes que realmente "gritam" (Log2FC alto ou baixo)
df_painel = df_assinatura[abs(df_assinatura['Log2FC']) > 1.5].sort_values('Log2FC')

# 6. EXIBIÇÃO DO PAINEL GÊNICO INTERATIVO
display(Markdown(f"# 📊 Painel de Vigilância Gênica"))
display(Markdown(f"Foram identificadas **{len(df_painel)} variantes críticas** para o fenótipo de vigilância."))

fig = px.bar(df_painel,
             x='Gene', y='Log2FC',
             color='Log2FC',
             color_continuous_scale='RdBu_r',
             title="Assinatura Gênica: Variantes de Expressão (Hiper-Vigilância vs Controle)",
             labels={'Log2FC': 'Força da Variante (Log2FC)'})

fig.update_layout(template='plotly_white', xaxis_tickangle=-45)
fig.show()

# Tabela resumo para o relatório
display(Markdown("### 📋 Top 10 Variantes Mineradas"))
display(df_painel.head(10))
```

```python
!pip install gseapy mygene pandas matplotlib seaborn

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import gseapy as gp
import mygene

# Dados de exemplo
probes = ["1553105_s_at","1553106_s_at","1552617_a_at","1553111_a_at",
          "1553219_a_at","1552977_a_at","1553530_a_at","117_at",
          "1553117_a_at","1553218_a_at"]
log2fc = [-9.005020,-7.637685,-7.594647,-7.321073,-6.860859,
          -6.304933,-6.185753,-6.165422,-5.869504,-5.729680]
pvals = [0.001,0.001,0.001,0.001,0.001,0.001,0.001,0.500,0.001,0.001]

df = pd.DataFrame({"Probe": probes, "Log2FC": log2fc, "p_valor": pvals})

# Mapear probes -> genes
mg = mygene.MyGeneInfo()
gene_info = mg.querymany(probes, scopes="reporter", fields="symbol,name", species="human")
gene_df = pd.DataFrame(gene_info)

df = df.merge(gene_df[["query","symbol","name"]], left_on="Probe", right_on="query", how="left")

# Separar reprimidos e ativados
df_down = df[df["Log2FC"] < 0]
df_up = df[df["Log2FC"] > 0]

# Inicializar variáveis
up_terms, down_terms = pd.DataFrame(), pd.DataFrame()

# Rodar enriquecimento apenas se houver genes ativados
if not df_up.empty:
    enr_up = gp.enrichr(gene_list=list(df_up["symbol"].dropna()),
                        gene_sets=["Reactome_2022","KEGG_2021_Human"],
                        organism="human")
    up_terms = enr_up.results[["Term","Adjusted P-value"]].head(10)

# Rodar enriquecimento apenas se houver genes reprimidos
if not df_down.empty:
    enr_down = gp.enrichr(gene_list=list(df_down["symbol"].dropna()),
                          gene_sets=["Reactome_2022","KEGG_2021_Human"],
                          organism="human")
    down_terms = enr_down.results[["Term","Adjusted P-value"]].head(10)

# Painel comparativo
plt.figure(figsize=(12,6))

if not up_terms.empty:
    sns.barplot(x="Term", y="Adjusted P-value", data=up_terms, color="red", label="Ativados (Imunidade)")

if not down_terms.empty:
    sns.barplot(x="Term", y="Adjusted P-value", data=down_terms, color="blue", label="Reprimidos (Manutenção)")

plt.xticks(rotation=45)
plt.legend()
plt.title("Comparação de vias: Repressão vs Ativação")
plt.show()
```
